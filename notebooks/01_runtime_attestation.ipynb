# QRAP Runtime Attestation Demo

This notebook demonstrates the Quantum Runtime Attestation Protocol (QRAP) core functionality including:
- Runtime initialization and lifecycle management
- Attestation generation and verification
- Integration with quantum simulations

```python
# Import the QRAP Python bindings
# Note: Build the Python package first with `maturin develop` in the python/ directory
try:
    import qrap_python as qrap
except ImportError:
    print("QRAP Python bindings not installed. Run 'maturin develop' in the python/ directory.")
    qrap = None
```

## 1. Runtime Configuration

The QRAP runtime can be configured with various parameters:

```python
if qrap:
    # Create a runtime configuration
    config = qrap.RuntimeConfig(
        num_qubits=4,
        max_execution_time_secs=60,
        verbose=True
    )
    
    print(f"Number of qubits: {config.num_qubits}")
    print(f"Verbose mode: {config.verbose}")
```

## 2. Runtime Lifecycle

The runtime follows a simple lifecycle: Ready → Running → Completed

```python
if qrap:
    # Create a runtime instance
    runtime = qrap.Runtime(config)
    print(f"Initial state: {runtime.state}")
    
    # Start the runtime
    runtime.start()
    print(f"After start: {runtime.state}")
    
    # Simulate some work
    import time
    time.sleep(0.1)
    
    # Stop the runtime
    elapsed = runtime.stop()
    print(f"After stop: {runtime.state}")
    print(f"Elapsed time: {elapsed:.4f} seconds")
    print(f"Execution count: {runtime.execution_count}")
```

## 3. Attestation Generation

Attestations provide cryptographic proof of quantum operations:

```python
if qrap:
    # Create an attestation for some data
    data = b"Quantum simulation result: success"
    attestation = qrap.Attestation(
        data=data,
        num_qubits=4,
        iterations=10,
        success=True
    )
    
    print(f"Attestation ID: {attestation.id}")
    print(f"Timestamp: {attestation.timestamp}")
    print(f"Data hash: {attestation.data_hash}")
    print(f"Signature: {attestation.signature[:32]}...")
    print(f"Num qubits: {attestation.num_qubits}")
    print(f"Iterations: {attestation.iterations}")
    print(f"Success: {attestation.success}")
```

## 4. Attestation Verification

```python
if qrap:
    # Verify the attestation
    is_valid = attestation.verify()
    print(f"Attestation is valid: {is_valid}")
    
    # Check expiry
    is_expired = attestation.is_expired(3600)  # 1 hour max age
    print(f"Attestation is expired (1h): {is_expired}")
    
    # Export as JSON
    json_str = attestation.to_json()
    print(f"\nJSON representation:\n{json_str}")
```
